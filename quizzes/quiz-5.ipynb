    ```python
    import numpy as np
    import os

    class Dataset:
        """Represents a dataset."""

        def __init__(self, filename):
            self.filename = filename

            with open(filename) as f:
                lines = f.readlines()
            
            rows = []
            for line in lines:
                row = []
                for x in line.split(","):
                    row.append(int(x.strip()))
                rows.append(row)
            self.data = np.array(rows)
    
        def __gt__(self, other):
            return np.sum(self.data) > np.sum(other.data)

    d1 = Dataset(os.path.join("data", "exam.txt"))
    d2 = Dataset(os.path.join("data", "exam2.txt"))
    print(np.sum(d1.data))
    print(np.sum(d2.data))
    print(d1 > d2)
    ```
