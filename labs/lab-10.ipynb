### Bellevue Almshouse Dataset

For this problem, we'll again work with the Bellevue Almshouse Dataset that was introduced in Lab 5.

In the code cell below, I've loaded in the dataset and converted it to a dictionary in the format required for the following questions. I'm using a lot of unfamilar code to do this – we'll be learning what all of it means in the future!

In the dictionary `bellevue_dict` the keys are people's names and the values are dictionaries storing the `date_in`, `age`, `disease`, `gender`, and `children` of each person.

*Note*: When creating the `bellevue_dict` object I have the code drop entries for people who have the same name as someone already in the dataset. This is solely for the purpose of this assignment – there are many reasons you would not do this when truly analyzing the data.

*Note*: I have also replaced missing values for the `children` field, which the data automatically labels `NaN` (or not-a-number) with empty strings, which will make downstream processing easier.

```python
import pandas as pd
import math

# Loads in Bellevue Almshouse Dataset and converts to dictionary
bellevue_df = pd.read_csv("lab-files/bellevue_almshouse_modified.csv")
bellevue_df["name"] = bellevue_df["first_name"] + " " + bellevue_df["last_name"]
bellevue_df = bellevue_df.drop(columns=["first_name", "last_name"])\
    .drop_duplicates(subset=["name"])\
    .set_index("name")
bellevue_df["children"] = bellevue_df["children"].fillna("")
bellevue_dict = bellevue_df.to_dict('index')
```

#### Part 5

Consider the code block below, in which I've written some analysis code using the `bellevue_dict` object. In 1-2 sentences, describe what the code does. Try to be as precise as possible!

*Note*: This code is intentionally complex! Don't worry if you struggle reading it or don't understand every detail.

```python
name_count = {}

for person in bellevue_dict:
    if not math.isnan(bellevue_dict[person]["age"]) and bellevue_dict[person]["age"] <= 30 and bellevue_dict[person]["age"] >= 20 and bellevue_dict[person]["children"]:
        last_name = person.replace("(", "").replace("?", "").replace(")", "").split()[-1]

        if last_name in name_count:
            name_count[last_name] += 1
        else:
            name_count[last_name] = 1

print(name_count)
```
