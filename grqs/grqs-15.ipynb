```python
import os

def count_letter(file_path, letter):
    with open(file_path) as f:
        text = f.read()
    
    count = 0
    for l in text:
        if l == letter:
            count += 1

    with open(file_path, "w") as f:
        f.write(f"There are {count} {letter}s in this text.")

count_letter(os.path.join("grqs-files", "test.txt"), "a")
count_letter(os.path.join("grqs-files", "test.txt"), "b")
```
