Two final tips might make interacting with the terminal easier for you:
- **Tab completion**: When typing in the terminal, if you hit the `tab` key, the command line will try to fill in what you're writing. For example, if I'm in my home directory and I type `cd Doc` then hit tab, it will fill in `cd Documents`.
- **Up and down arrows**: The up and down arrows can be used to go through the commands you've already run. Hit the up arrow once and the last command you ran will appear before your cursor, and so on.
