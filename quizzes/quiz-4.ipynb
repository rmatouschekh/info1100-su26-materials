    ```python
    import re

    class Course:
        """Represents a course."""

    def make_course(name, number, students):
        c = Course()
        c.name = name
        c.number = number
        c.students = students
        return c

    def vowel_students(course):
        result = set()
        for name in course.students:
            if re.search(r"^[AEIOU]", name):
                result.add(name)
        return result

    c = make_course("Data Structures", 2110,
                    ["Agatha Christie", "Alexandre Dumas", "Marcel Proust", "Agatha Christie"])
    found = vowel_students(c)
    print(len(found))
    print("Marcel Proust" in found)
    print("Agatha Christie" in found)
    ```

    ```python
    import re

    class Course:
        """Represents a course."""

    def make_course(name, number, students):
        c = Course()
        c.name = name
        c.number = number
        c.students = students

    def course_tag(course):
        words = re.findall(r"[A-Z][a-z]+", course.name)
        return (words[0], course.number)

    def relabel(tag, new_num):

        return tag

    c = make_course("Introduction to Programming", 1100, ["Jane Austen"])
    tag = course_tag(c)
    print(tag)

    tag = relabel(tag, 2200)
    ```
