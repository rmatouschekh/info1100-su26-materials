### Checking Membership

Below, I've loaded a list of valid scrabble words and the copy of *Pride and Prejudice* by Jane Austen available from Project Gutenberg. The scrabble words are stored as a list of lowercase words in `scrabble_words`. I've also removed all punctuation from the copy of *Pride and Prejudice* and transformed it into a list of lowercase words stored in `book_words`.

```python
import os
import string

# Loading scrabble words
with open(os.path.join("lab-files", "scrabble_words.txt")) as f:
    scrabble_words = [w.strip().lower() for w in f.readlines()]

# Loading book words
with open(os.path.join("lab-files", "pride_and_prejudice.txt")) as f:
    book = f.read()
translator = str.maketrans({c: " " for c in string.punctuation + '”“’…'})
book_words = book.lower().translate(translator).split()
```

The code below counts and then prints the number of words in *Pride and Prejudice* that are not in the list of Scrabble words. Some extra code times how long the code takes – for me, it says 3 minutes and 8 seconds, or about 188.1 seconds total! You can rerun the code to see how long it takes on your machine, or just use mine as reference.

```python
import time

start_time = time.perf_counter()

not_scrabble_words = 0

for word in book_words:
    if word not in scrabble_words:
        not_scrabble_words += 1

print(f"There are {not_scrabble_words} non-Scrabble words in Pride and Prejudice.")

end_time = time.perf_counter()
print(f"Elapsed time: {end_time-start_time:.4f} seconds")
```
