### Adding to Sorted Lists

*Adapted from Cornell's CS 1110 (Spring 2025) materials*

Consider the following set of list functions, most of which the readings for today introduced:

| Method | Description |
|--------|-------------|
| `x.index(c)` | **Returns**: the first position of `c` in list `x`; error if not there |
| `x.count(c)` | **Returns**: the number of times that `c` appears in the list `x` |
| `x.append(c)` | Adds the value `c` to the end of the list. This method **alters** the list; it does not make a new list |
| `x.insert(p,c)` | Inserts the value `c` at position `p`, shifting all other elements to the right. This method **alters** the list; it does not make a new list |
| `x.sort()` | Sorts the contents of the list to be in ascending order (type dependent). This method **alters** the list; it does not make a new list |
| `x.pop()` | Removes and **returns** the last element of the list. This method **alters** the list; it does not make a new list |
